```python
#PART 1

def f(x):
    return (x**3) - (x**2) - 1

def df(x):
    return 3*(x**2) - 2*(x)
```

```python
#PART 2

def newton(f, df, x0, epsilon=1e-6, max_iter=30):
    xn = x0

    for n in range(1, max_iter + 1):
        f_xn = f(xn)
        df_xn = df(xn)
        
        if abs(f_xn) < epsilon:
            print(f"Found root in {n-1} iterations")
            return xn

        if df_xn == 0:
            print("Iteration failed")
            return None

        xn = xn - (f_xn/df_xn)

    print("Iteration failed")
    return None
```

```python
# PART 3
x0_1 = 2
x0_2 = 5

print(newton(f, df, x0_1))
print(newton(f, df, x0_2))

epsilon=1e-8
print(/n"Now with an epsilon value of 1e-8")
print(newton(f, df, x0_1, epsilon))
print(newton(f, df, x0_2, epsilon))
```
